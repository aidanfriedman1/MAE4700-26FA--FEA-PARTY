```python
import numpy as np
import matplotlib.pyplot as plt
```

We will first implement the code to define our exact displacement and strain fields, knowing strain = d/dx(displacement).

```python

# Exact displacement field
def u_exact(x):
    return x**3

# Exact strain field
def strain_exact(x):
    return 3*x**2
```

Part 3a: Approximation with 2 elements

```python

# Define nodes for two-element approximation
nodes = np.array([0, 0.5, 1.0])

# Calculate nodal displacements
u_nodes = u_exact(nodes)

# Points for plotting the FE displacement field
x_FE = []
u_FE = []

# Loop through elements one at a time
for e in range(len(nodes) - 1):

    # Nodes belonging to element e
    x1 = nodes[e]
    x2 = nodes[e + 1]

    # Element length
    h_e = x2 - x1

    # Element nodal displacement vector
    d_e = np.array([u_nodes[e], u_nodes[e + 1]])

    # x values within the element
    x_e = np.linspace(x1, x2, 50)

    # Linear shape functions
    N1 = (x2 - x_e) / h_e
    N2 = (x_e - x1) / h_e

    # Displacement field: u^e(x) = N^e(x)d^e
    u_e = N1*d_e[0] + N2*d_e[1]

    # Store values for plotting
    x_FE.extend(x_e)
    u_FE.extend(u_e)
```

Plotting u(x) and finite element field on [0,1]:

```python

# Exact solution on [0,1]
x = np.linspace(0, 1, 1000)

plt.figure()

# Exact displacement field
plt.plot(x, u_exact(x), label='Exact: $u(x)=x^3$')

# Finite element displacement field
plt.plot(x_FE, u_FE, label='Finite Element')

# Show nodes
plt.scatter(nodes, u_nodes, label='Nodes')

plt.xlabel('x')
plt.ylabel('u(x)')
plt.title('Exact vs. Finite Element Displacement Field')
plt.legend()
plt.grid()

plt.show()
```

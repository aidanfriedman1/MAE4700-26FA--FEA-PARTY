We will first implement the code to define our exact displacement and strain fields, knowing that strain = d/dx(displacement).

```python

# Exact displacement field - given in problem statement
def u_exact(x):
    return x**3

# Exact strain field
def strain_exact(x):
    return 3*x**2
```

Part 3a: Displacement approximation with 2 elements

```python

# Define nodes for two-element approximation
nodes = np.array([0, 0.5, 1.0])

# Calculate nodal displacements
u_nodes = u_exact(nodes)

# Points for plotting the FE displacement field
x_FE = []
u_FE = []

# Loop through elements one at a time
for e in range(len(nodes) - 1):

    # Nodes belonging to element e
    x1 = nodes[e]
    x2 = nodes[e + 1]

    # Element length
    h_e = x2 - x1

    # Element nodal displacement vector
    d_e = np.array([u_nodes[e], u_nodes[e + 1]])

    # x values within the element
    x_e = np.linspace(x1, x2, 50)

    # Implementing linear shape functions
    N1 = (x2 - x_e) / h_e
    N2 = (x_e - x1) / h_e

    # Implementing displacement field: u^e(x) = N^e(x)d^e
    u_e = N1*d_e[0] + N2*d_e[1]

    # Store values for plotting
    x_FE.extend(x_e)
    u_FE.extend(u_e)
```

Plotting u(x) and finite element field on [0,1]:

```python

# Exact solution on [0,1]
x = np.linspace(0, 1, 1000)

plt.figure()

# Exact displacement field
plt.plot(x, u_exact(x), label='Exact: $u(x)=x^3$')

# Finite element displacement field
plt.plot(x_FE, u_FE, label='Finite Element')

# Plotting nodes
plt.scatter(nodes, u_nodes, label='Nodes')

plt.xlabel('x')
plt.ylabel('u(x)')
plt.title('Exact vs. Finite Element Displacement Field')
plt.legend()
plt.grid()

plt.show()
```

Part 3b: Strain finite element computation

```python
# Arrays to store strain
x_strain_FE = []
strain_FE = []

# Looping through each element one at a time
for e in range(len(nodes) - 1):

    # Element nodes
    x1 = nodes[e]
    x2 = nodes[e + 1]

    # Element length
    h_e = x2 - x1

    # Element nodal displacement vector
    d_e = np.array([u_nodes[e], u_nodes[e + 1]])

    # B matrix
    B = np.array([-1/h_e, 1/h_e])

    # Element strain: epsilon^e = B^e d^e
    strain_e = B @ d_e

    # Points for plotting
    x_e = np.linspace(x1, x2, 100)

    # Store values
    x_strain_FE.extend(x_e)
    strain_FE.extend(np.ones(len(x_e)) * strain_e)
```

Plotting finite element strain alongside the exact strain:

```python
# Exact solution points
x = np.linspace(0, 1, 500)

plt.figure()

plt.plot(x, strain_exact(x), label='Exact Strain')
plt.plot(x_strain_FE, strain_FE, label='Finite Element Strain')

plt.xlabel('x')
plt.ylabel('Strain')
plt.title('Exact vs. Finite Element Strain')
plt.legend()
plt.grid()

plt.show()
```

Discussion: Does the interpolation of the strain improve?

Yes, it appears that the interpolation of the strain does improve as we increase the number of elements. Because we are using linear shape functions as given in the lecture notes to represent our displacement field, we see that the finite element strain is constant within each element. Therefore, the finite element strain as plotted above resembles a step function whereas the exact strain varies continously with x. As we increase our number of elements from 2 to 4 and then to 8, we are effectively reducing the length of each element, which results in the finite element strain field more closely matching the exact strain field. Overall, the error that we get from our finite element strain approximation decreases as we increase the number of elements over the same region, [0,1] in this case.
